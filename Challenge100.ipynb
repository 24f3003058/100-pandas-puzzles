# Dataframe Basics

```python
import numpy as np
data={
    'fruits':['apple','orange','strawberry','cherry','banana'],
    'color':['red','orange','reddish','pinkish-red','yellowish-white'],
    'price':[200,150,250,120,60]
}

labels=[1,2,3,4,5]

df=pd.DataFrame(data, index=labels)
print(df)
```

```python
df.info()
```

```python
df.describe()
```

```python
df.head()
```

```python
df.loc[:,['price','fruits']]
```

```python
df[df['price'] >=150]
```

```python
df[df['price'].isnull()]
```

```python
df[(df['color']=='red') | (df['price']>=150)]
```

```python
df[df['price'].between(100,222)]
```
