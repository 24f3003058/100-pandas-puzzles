# Dataframe Basics

```python
import numpy as np
data={
    'fruits':['apple','orange','strawberry','cherry','banana'],
    'color':['red','orange','reddish','pinkish-red','yellowish-white'],
    'price':[200,150,250,120,60]
}

labels=[1,2,3,4,5]

df=pd.DataFrame(data, index=labels)
print(df)
```
